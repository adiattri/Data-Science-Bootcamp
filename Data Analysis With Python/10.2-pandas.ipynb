```python
### Series

import pandas as pd
```

```python
data=[1,2,3,4,5]
series=pd.Series(data)
print(series)
```

```python
### Create a series from dictionary elements

data = {'a':1,'b':2,'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10,20,30]

index=['a','b','c']

pd.Series(data,index)
```

```python
### Data Frame

data = {
    'Name':['Krish', 'John', 'Jack'],
    'Age': [20,30,45],
    'City': ['Banglore', 'New york', 'Florida']
}
df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```
