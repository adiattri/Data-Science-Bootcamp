-Simple linear regression

A model takes weight and predicts the height. Weight is independent feature and height is dependent feature. We create a best fit line such as whenever we give a new weight the model is able to predict the height.

Notations to be used: 

equation to be used: h(theta)x = (theta)0 + (theta)1x
what is theta 0 = it is the intercept
what is theta 1 = it is called slope or coefficient
