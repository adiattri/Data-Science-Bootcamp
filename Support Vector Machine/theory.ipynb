This machine learning algorithm is used to solve both classification and regression problem.


When solving a classification problem: Support vector classifier (SVC)
When solving a regression problem: Support Vector Regressor


1 - SVC
    Along with best fit line we also try to create two more lines which are called as marginal planes. The marginal plane should be created in such a way that distance between them should be maximum. Means only that best fit line will be choosen with max marginal plane distance. The point thorugh which planes pass are called support vectors.

    3D - two marginal planes will be created along with the best fit plane.
    
    distance between marginal planes = 2/absolute w

    cost function of SVC minimize ||w||/2 + Ci summation etaa of i (Hinge loss)

    what is Ci?
    -how many points we consider to avoid misclssification


    what is eetaaa?
    -summation of the distance of the incorrect data points from the marginal plane
